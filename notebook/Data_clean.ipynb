# Credit Card Default Prediction - Data Cleaning and Initial EDA

This notebook checks and explores the credit card customer data provided for the Inter-Uni Datathon.

The aim is to understand the structure and quality of the data, identify any cleaning that is required, and explore the main variables that may be related to customer default before building prediction models.


```python
import numpy as np
import pandas as pd

train = pd.read_csv("../data/raw/train.csv")
test = pd.read_csv("../data/raw/test.csv")
```

Check the size and structure of the training and test datasets to ensure they have loaded correctly.

```python
print("Train shape:", train.shape)
print("Test shape:", test.shape)
```

```python
display(train.head())
display(test.head())
```

```python
print("Train columns:")
print(train.columns.tolist())

print("\nTest columns:")
print(test.columns.tolist())
```

Confirm that default is only present in the training set.

```python
set(train.columns) - set(test.columns)
```

```python
train.dtypes
```

Check for missing data

```python
print("Missing values in train:", train.isna().sum().sum())
print("Missing values in test:", test.isna().sum().sum())
```

Check for duplicates

```python
print("Duplicate rows in train:", train.duplicated().sum())
print("Duplicate rows in test:", test.duplicated().sum())
```

Check for duplicate ids

```python
print("Duplicate client IDs in train:", train["client_id"].duplicated().sum())
print("Duplicate client IDs in test:", test["client_id"].duplicated().sum())
```

Get basic numerical summary

```python
train.describe().T
```

Columns such as `SEX`, `EDUCATION` and `MARRIAGE` are stored as numerical codes rather than true numerical values.

We check the unique category values and their counts to identify any unexpected or undocumented values.

```python
categorical_cols = ["SEX", "EDUCATION", "MARRIAGE"]

for col in categorical_cols:
    print(f"\n{col}")
    print(train[col].value_counts().sort_index())
```

Check the values present in the repayment status columns before analysis.

```python
pay = ["PAY_0", "PAY_2", "PAY_3", "PAY_4", "PAY_5", "PAY_6"]

for col in pay:
    print(f"\n{col}")
    print(train[col].value_counts().sort_index())
```
