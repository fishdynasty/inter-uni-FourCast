```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

train = pd.read_csv("../data/raw/train.csv")
test = pd.read_csv("../data/raw/test.csv")

# 1. 目標欄位分佈 —— 先看違約 vs 沒違約的比例
train['default'].value_counts(normalize=True)
sns.countplot(x='default', data=train)
```

```python
# 2. 數值特徵基本統計
train.describe()

# 3. 看年齡、額度分佈
train[['AGE', 'LIMIT_BAL']].hist(bins=30, figsize=(10,4))

# 4. 各特徵跟 default 的關聯
corr = train.drop(columns=['client_id']).corr()
plt.figure(figsize=(12,10))
sns.heatmap(corr, cmap='coolwarm', center=0)
```

```python
# 5. 重點特徵單獨看跟 default 的關係
plt.figure()
sns.boxplot(x='default', y='LIMIT_BAL', data=train)
plt.title('LIMIT_BAL by default')

plt.figure()
sns.countplot(x='PAY_0', hue='default', data=train)
plt.title('PAY_0 by default')
```
